```python
# Import và Đọc dữ liệu
import pandas as pd

routes = pd.read_csv('raw_data/routes.txt')
stops = pd.read_csv('raw_data/stops.txt')
trips = pd.read_csv('raw_data/trips.txt')
stop_times = pd.read_csv('raw_data/stop_times.txt')
```

```python
# Xử lý Logic
metro_routes = routes[routes['route_type'] == 1]
metro_trips = trips[trips['route_id'].isin(metro_routes['route_id'])]
metro_stop_times = stop_times[stop_times['trip_id'].isin(metro_trips['trip_id'])].copy()
metro_stops = stops[stops['stop_id'].isin(metro_stop_times['stop_id'])].copy()

metro_stop_times = metro_stop_times.dropna(subset=['arrival_time', 'departure_time'])

def fix_time(time_str):
    h, m, s = map(int, str(time_str).split(':'))
    if h >= 24:
        h = h - 24
    return f"{h:02d}:{m:02d}:{s:02d}"

metro_stop_times['arrival_time'] = metro_stop_times['arrival_time'].apply(fix_time)
metro_stop_times['departure_time'] = metro_stop_times['departure_time'].apply(fix_time)
```

```python
# Dọn dẹp và Lưu dữ liệu
import os

# Tạo thư mục nếu lỡ tay xóa mất
output_dir = 'processed_data'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Chuyển đổi định dạng
metro_stop_times['arrival_time'] = pd.to_timedelta(metro_stop_times['arrival_time'])
metro_stop_times['departure_time'] = pd.to_timedelta(metro_stop_times['departure_time'])
metro_stops['lat_lon'] = metro_stops['stop_lat'].round(4).astype(str) + '_' + metro_stops['stop_lon'].round(4).astype(str)

# Lưu vào thư mục processed_data
metro_stops.to_csv(os.path.join(output_dir, 'cleaned_stops.csv'), index=False)
metro_stop_times.to_csv(os.path.join(output_dir, 'cleaned_stop_times.csv'), index=False)

print(f"Đã dọn dẹp và lưu file vào thư mục '{output_dir}' thành công!")
```
